## Build Global PCA Basis

The counterfactual latent-space figure uses a global PCA basis fit on the
Deep4Chem training embedding space. Run this once to regenerate the basis if

```python
!{sys.executable} scripts/analysis/build_global_pca_basis.py \
  --train-emb local_runs/counterfactuals_cache/train.parquet \
  --train-csv data/uvvisml/abs/deep4chem/splits/scaffold/smiles_target_train.csv \
  --ckpt checkpoints/abs/deep4chem/scaffold/morgan_fingerprint/fromscratch_N11816_s42/fold_0/model_0/model.pt \
  --device cpu \
  --outdir figures/latent_interpretation \
  --basis-name global_pca_basis.joblib \
  --curve-name global_pca_basis_cumvar.pdf
```

## Generate Counterfactual Figure

This script regenerates the reported counterfactual latent-space figure and the

```python
!{sys.executable} scripts/analysis/plot_counterfactuals.py \
  --ckpt-dir checkpoints/abs/deep4chem/scaffold/morgan_fingerprint/fromscratch_N11816_s42/fold_0 \
  --pca-basis figures/latent_interpretation/global_pca_basis.joblib \
  --device cpu \
  --outdir figures/molecules/counterfactuals
```
