# Dataset Similarities (Trimmed)

This notebook is a lightweight front end for the cleaned dataset-similarity workflow. The canonical figure-generation logic lives in script entry points under `scripts/analysis/`.

## Figure Pipeline

The main paper uses two figure families from this analysis:

- a combined dataset UMAP showing the relative chemical-space coverage of selected datasets
- split-similarity CDFs showing how scaffold and random test sets compare to their train sets

Those are generated by:

- `scripts/analysis/plot_dataset_umap.py`
- `scripts/analysis/plot_split_similarity_cdfs.py`


```python
from pathlib import Path
import os
import sys

cwd = Path.cwd().resolve()
REPO_ROOT = cwd.parent if cwd.name == "notebooks" else cwd
os.chdir(REPO_ROOT)

print(f"Repo root: {REPO_ROOT}")
print(f"Python: {sys.executable}")
```

```python
!{sys.executable} scripts/analysis/plot_dataset_umap.py \
  --dataset-dir deep4chem=data/uvvisml/abs/deep4chem/splits/scaffold/ \
  --dataset-dir chemfluor=data/uvvisml/abs/chemfluor/splits/scaffold/ \
  --dataset-dir dsscdb=data/uvvisml/abs/dsscdb/splits/scaffold/ \
  --dataset-dir jeffries=data/custom/em/jeffries/splits/holdout \
  --outdir figures/datasets \
  --fig-name dataset_umap.pdf \
  --source-data-name dataset_umap_source_data.csv
```

The UMAP script writes:

- `figures/datasets/dataset_umap.pdf`
- `figures/datasets/dataset_umap_source_data.csv`


```python
!{sys.executable} scripts/analysis/plot_split_similarity_cdfs.py \
  --comparison Deep4Chem=data/uvvisml/abs/deep4chem/splits/scaffold=data/uvvisml/abs/deep4chem/splits/random \
  --comparison ChemFluor=data/uvvisml/abs/chemfluor/splits/scaffold=data/uvvisml/abs/chemfluor/splits/random \
  --comparison DSSCDB=data/uvvisml/abs/dsscdb/splits/scaffold=data/uvvisml/abs/dsscdb/splits/random \
  --outdir figures/datasets \
  --fig-name split_similarity_cdfs.pdf \
  --source-data-name split_similarity_cdfs_source_data.csv
```

The CDF script writes:

- `figures/datasets/split_similarity_cdfs.pdf`
- `figures/datasets/split_similarity_cdfs_source_data.csv`


## Notes

- The UMAP figure accepts arbitrary `label=split_dir` inputs, so Jeffries can be included explicitly without hardcoding it into the script defaults.
- The CDF script is organized around paired scaffold/random comparisons for the same dataset.
