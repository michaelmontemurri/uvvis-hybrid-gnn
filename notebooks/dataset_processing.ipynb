# Dataset Processing Provenance

This notebook is a short provenance record for the custom Jeffries dataset used in the repo.
The canonical build logic now lives in `scripts/build_jeffries_dataset.py`.


## What The Build Script Does

Starting from `data/raw/jeffries.xlsx`, the script:

- keeps the 58 worksheet rows used in the study,
- reshapes the solvent-specific emission columns into long format,
- renames columns into the repo's standard schema,
- applies the manual CHCl3 corrections for the four designated holdout molecules,
- maps solvent labels to solvent SMILES,
- generates 256-bit solvent Morgan fingerprints,
- filters rows with missing or implausibly small emission values, and
- writes the processed Jeffries tables under `data/custom/{target}/jeffries/`.


## Manual Corrections Captured In The Script

The following CHCl3 emission values are explicitly patched before writing the processed dataset:

- `26Ad48C2` -> `414`
- `26Ad48C3` -> `410`
- `26P48C2` -> `470`
- `26P48C3` -> `454`

These corrections were part of the original notebook workflow and are now encoded directly in the script.
